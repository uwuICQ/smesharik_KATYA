### Задача 1. Среднее, медиана, мода

Напиши функции `mean`, `median` и `mode`. Проверь их на двух рядах и выведи три числа для каждого.

- ряд A: `320, 340, 350, 360, 360, 380, 400, 2100`
- ряд B: `3, 4, 4, 5, 4, 3, 5, 4`


```python
from math import *

def mean(spisok):
    if len(spisok) != 0:
        return sum(spisok)/len(spisok)
    else: return 0

def median(spisok):
    if len(spisok) != 0:
        cur = len(spisok)/2
        if cur == int(cur): 
            return spisok[int(cur)]
        else :
            return (spisok[int(cur)] + spisok[ceil(cur)])/2
        
    else: return 0

def mode(spisok):
    maa = max([spisok.count(x) for x in spisok])
    freq = [x for x in spisok if spisok.count(x)==maa]
    return freq[0]

A = [320, 340, 350, 360, 360, 380, 400, 2100]
B = [3, 4, 4, 5, 4, 3, 5, 4]

print('Для А: ', mean(A), median(A), mode(A))
print('Для B: ', mean(B), median(B), mode(B))
```

### Задача 2. Размах, дисперсия, стандартное отклонение

Ряд: `2, 4, 4, 4, 5, 5, 7, 9`.

Напиши `data_range`, `variance` и `std`. Выведи три числа для исходного ряда, для ряда «к каждому элементу прибавить 10» и для ряда «каждый элемент умножить на 2».

По выводу должно быть видно свойство: сдвиг дисперсию не меняет, удвоение каждого числа умножает дисперсию на 4, а стандартное отклонение на 2.


```python
def data_range(spisok):
    return max(spisok) - min(spisok)

#для дисперсии нужно среднее, приведу прогу из предыдущего задания
def mean(spisok):
    if len(spisok) != 0:
        return sum(spisok)/len(spisok)
    else: return 0

def variance(spisok):
    sred = mean(spisok)
    summa = sum([(x - sred)**2 for x in spisok])
    return summa/len(spisok)

def std(spisok):
    sred = mean(spisok)
    summa = sum([(x - sred)**2 for x in spisok])
    return (summa/len(spisok))** 0.5

A = [2, 4, 4, 4, 5, 5, 7, 9]
A1 = [x+10 for x in A]
A2 = [x*2 for x in A]


print('1 запрос: ', data_range(A), variance(A), std(A))
print('2 запрос: ', data_range(A1), variance(A1), std(A1))
print('3 запрос: ', data_range(A2), variance(A2), std(A2))
```

### Задача 3. Корреляция Пирсона по списку словарей

Формула, её нужно написать самой:

`r = сумма (x − среднее x) * (y − среднее y) / корень( сумма (x − среднее x)^2 * сумма (y − среднее y)^2 )`

Заказы кафе. Каждый элемент — словарь с ключами `day`, `bill`, `tip`.

| day | bill | tip |
| --- | --- | --- |
| пн | 400 | 40 |
| пн | 450 | 50 |
| пн | 420 | 30 |
| вт | 500 | 70 |
| вт | 900 | 120 |
| вт | 480 | 60 |
| ср | 350 | 40 |
| ср | 360 | 35 |
| ср | 2000 | 50 |
| чт | 600 | 90 |
| чт | 620 | 100 |
| чт | 580 | 80 |

Собери из списка словарей два списка: суммы чеков и чаевые. Напиши функцию `pearson`. Выведи коэффициент, округлённый до двух знаков через `round(..., 2)`.


```python
#для функции нужно среднее, приведу прогу из первого задания
def mean(spisok):
    if len(spisok) != 0:
        return sum(spisok)/len(spisok)
    else: return 0

def pearson(X, Y):
    xsred, ysred = mean(X), mean(Y)
    xysum = sum([(i - xsred) * (z - ysred) for i,z in zip(X,Y)])
    sqxsum = sum([(i - xsred)**2 for i in X])
    sqysum = sum([(i - ysred)**2 for i in Y])
    return xysum / ((sqxsum * sqysum) ** 0.5)
    
bill = [400, 450, 420, 500, 900, 480, 350, 360, 2000, 600, 620, 580]
tip = [40, 50, 30, 70, 120, 60, 40, 35, 50, 90, 100, 80]
print(round(pearson(bill, tip), 2))
```

## 6. График как в лекции

Те же двенадцать заказов, что в задаче 3.

В ноутбуке лекции по разведочному анализу на одном наборе рисуют три разные картинки и читают каждую отдельно. Сделай так же, matplotlib или seaborn. На каждом графике нужны заголовок и подписи осей. Данные можно держать списками или таблицей pandas.

1. Гистограмма `bill`: где суммы густые и есть ли хвост.
2. Ящик с усами: `bill` в разрезе дня.
3. Диаграмма рассеяния: по оси x сумма чека `bill`, по оси y чаевые `tip`.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# сгенерила набор синтетических данных для простроения графиков

rng = np.random.default_rng(42)
n = 200

days = rng.choice(["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Вс"], size=n,
p=[0.08, 0.08, 0.10, 0.12, 0.18, 0.24, 0.20])

bill = rng.lognormal(mean=3.0, sigma=0.5, size=n)
bill = np.round(bill, 2)

weekend = np.isin(days, ["Сб", "Вс"])
bill[weekend] = np.round(bill[weekend] * 1.25, 2)

tip_rate = rng.normal(loc=0.15, scale=0.04, size=n).clip(0.02, 0.30)
tip = np.round(bill * tip_rate, 2)

df = pd.DataFrame({"day": days, "bill": bill, "tip": tip})

# гистограмма

plt.hist(df["bill"], bins = 7)
plt.title("распределение сумм чеков")
plt.xlabel("Сумма чеков")
plt.ylabel("Количество чеков")
plt.show

# тепрерь ящик с усами 
df.boxplot(column="bill", by="day")
plt.title("Сумма чеков по дням недели")
plt.xlabel("День недели ")
plt.ylabel("Сумма чека")
plt.show

# сделала диаграмму рассеивания в другой ячейке. ниже


```
